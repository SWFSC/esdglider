# Examples of using the QARTOD module

## Create a new NetCDF with QARTOD flags

```python
from esdglider.qartod import run_qartod_qc

run_qartod_qc(
    input_file="/Users/madisonrichardson/Desktop/esdglider/sam-nc-files/amlr08-20220513/non-qc/amlr08-20220513T1858.nc",
    output_file="/Users/madisonrichardson/Desktop/esdglider/sam-nc-files/amlr08-20220513/qc/amlr08-20220513T1858_qc.nc",
    config_file="/Users/madisonrichardson/Desktop/esdglider-madi/esdglider/data/qartod-config.yml"
)
```

## Update an existing NetCDF file

If a NetCDf already contains placeholder QC variables (or if you want to overwrite the file) the same file can be used as both input and output.

```python
from esdglider.qartod import run_qartod_qc

run_qartod_qc(
    input_file="/Users/madisonrichardson/Desktop/esdglider/sam-nc-files/amlr08-20220513/non-qc/amlr08-20220513T2023.nc",
    output_file="/Users/madisonrichardson/Desktop/esdglider/sam-nc-files/amlr08-20220513/non-qc/amlr08-20220513T2023.nc",
    config_file="/Users/madisonrichardson/Desktop/esdglider-madi/esdglider/data/qartod-config.yml",
    overwrite_qc=True
)
```
