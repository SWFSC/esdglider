# Running the QARTOD workflow on a full deployment and creating plots + CSV file

## Import packages

```python
from pathlib import Path

from esdglider.plots import (
    plot_qc_summary,
    plot_qc_timeseries,
)
from esdglider.qartod import (
    create_qc_summary_table,
    run_qartod_qc,
)
from esdglider.utils import combine_datasets
```

## Run QARTOD QC on every profile in the amlr08-20220513 deployment

```python
# Input directory
input_dir = Path(
    "/Users/madisonrichardson/Desktop/esdglider/"
    "sam-nc-files/amlr08-20220513/"
    "processed-L1/ngdac/delayed/non-qc"
)

# Output directory
output_dir = Path(
    "/Users/madisonrichardson/Desktop/esdglider/"
    "sam-nc-files/amlr08-20220513/"
    "processed-L1/ngdac/delayed/qc"
)

# Location of qartod-config.yml
config_file = (
    "/Users/madisonrichardson/Desktop/esdglider-madi/"
    "esdglider/data/qartod-config.yml"
)

# Make sure output directory exists
output_dir.mkdir(
    parents=True,
    exist_ok=True,
)

# Process each netcdf file
for input_file in sorted(input_dir.glob("*.nc")):
    # Build output file name
    output_file = (
        output_dir
        / f"{input_file.stem}_qc.nc"
    )
    print(f"Processing {input_file.name}")
    try:
        # Run complete QARTOD workflow
        run_qartod_qc(
            input_file=str(input_file),
            output_file=str(output_file),
            config_file=config_file,
        )
    # Report files that fail without stopping workflow
    except Exception as e:
        print(f"Failed: {input_file.name}")
        print(e)

print("Done!")
```

* This error: **Could not run "qartod.rate_of_change_test: rate_of_change_test() missing 1 required positional argument: 'tinp'** is only because of trying to run this test on variable u_qc and v_qc, which can be disabled in the qartod-config.yml

## Create deployment figures and CSV file

```python
# Directory of _qc.nc files
qc_dir = output_dir

# Output directory for plots
plot_dir = (
    output_dir.parent
    / "plots"
)
plot_dir.mkdir(exist_ok=True)

# Combine all profile nc files into single deployment dataset
print("Combining deployment QC datasets...")
deployment_ds = combine_datasets(
    qc_dir,
)

# Create summary table of all qc flags for deployment
print("Creating deployment QC summary table...")
profile_summary_df = create_qc_summary_table(
    deployment_ds,
)

# Create summary plot of all qc flags for deployment
print("Creating deployment QC summary plot...")
summary_df = plot_qc_summary(
    deployment_ds,
    plot_file=(
        plot_dir
        / "amlr08-20220513-deployment_qc_summary.png"
    ),
)

# Create a CSV with profile name, profile index, and qc flags
print("Writing deployment QC summary CSV...")
profile_summary_df.to_csv(
    plot_dir
    / "amlr08-20220513-deployment_qc_summary.csv",
    index=False,
)

# Create timeseries plots of qc flags for every variable with time dim
print("Creating QC timeseries plots...")
plot_qc_timeseries(
    deployment_ds,
    output_dir=plot_dir,
    deployment_name="amlr08-20220513 Deployment",
)

print("\nFinished processing deployment.")
print(
    f"Summary plot saved to:\n"
    f"{plot_dir / 'amlr08-20220513-deployment_qc_summary.png'}"
)
print(
    f"Summary CSV saved to:\n"
    f"{plot_dir / 'amlr08-20220513-deployment_qc_summary.csv'}"
)
print(
    f"Timeseries plots saved to:\n"
    f"{plot_dir}"
)
print(
    f"\nProcessed {len(summary_df)} QC variables "
    f"across {len(profile_summary_df)} profiles."
)
```
